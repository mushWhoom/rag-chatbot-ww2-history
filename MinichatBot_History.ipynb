## **Design Sistem RAG**

###**1. Model Embedding**

Komputer pada dasarnya tidak mengerti bahasa manusia. Jadi, model ini bertugas menerjemahkan kalimat-kalimat dari PDF yang saya pakai menjadi susunan vektor angka.

###**2. Vector Database (Chroma)**
 Ini adalah tempat penyimpanan khusus untuk menampung semua vektor angka tadi. Ketika kita memasukkan pertanyaan, database ini akan menghitung jarak antar vektor untuk mencari data mana yang posisinya paling dekat (paling relevan) dengan pertanyaan.

###**3. Model Bahasa / LLM (Gemini)**
 Setelah database menemukan datanya, LLM bertugas mengambil informasi kaku tersebut dan merangkainya kembali menjadi kalimat bahasa Indonesia yang santai seperti sedang diajak ngobrol.

## **Implementasi Model RAG**

```python
!pip install -q sentence-transformers
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_community.vectorstores import Chroma
from langchain_community.embeddings import HuggingFaceEmbeddings
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.runnables import RunnablePassthrough
from langchain_core.output_parsers import StrOutputParser

# untuk memotong teks dari file yang diupload (pdf)
text_splitter = RecursiveCharacterTextSplitter(chunk_size=1000, chunk_overlap=100)
chunks = text_splitter.split_documents(pages)

# model embeddingnya pake yang versi multilingual agar bisa bahasa indonesia
embeddings = HuggingFaceEmbeddings(model_name="sentence-transformers/paraphrase-multilingual-MiniLM-L12-v2")

# vector database dan menambahkan parameter persist_directory untuk menyimpan database ke folder lokal
vectorstore = Chroma.from_documents(
    documents=chunks,
    embedding=embeddings,
    persist_directory="./chroma_db_pd2"
)

print("✅ Vector Database berhasil disimpan ke folder './chroma_db_pd2'!")

# model bahasa llm nya pake yang terbaru
llm = ChatGoogleGenerativeAI(model="gemini-2.5-flash")

# pengimplementasian sistem RAGnya dengan template yang rapih ke bawah
retriever = vectorstore.as_retriever()

template = """Gunakan potongan konteks berikut untuk menjawab pertanyaan.
Jika kamu tidak tahu jawabannya, katakan saja bahwa kamu tidak tahu.

Instruksi Tambahan: Tolong jabarkan jawabanmu menggunakan poin-poin (bullet points) agar rapi dan mudah dibaca.

Konteks: {context}

Pertanyaan: {question}
"""
prompt = ChatPromptTemplate.from_template(template)

def format_docs(docs):
    return "\n\n".join(doc.page_content for doc in docs)

chatbot_pd2 = (
    {"context": retriever | format_docs, "question": RunnablePassthrough()}
    | prompt
    | llm
    | StrOutputParser()
)

print("AHAYYYYYYY! Mesin Aplikasi 'Chatbot RAG Polyglot' sudah siap!")
```

```python
from langchain_community.vectorstores import Chroma

# Memuat (load) kembali vector database dari direktori yang sudah disimpan
loaded_vectorstore = Chroma(
    persist_directory="./chroma_db_pd2",
    embedding_function=embeddings
)

print("Vector Database berhasil dimuat dari penyimpanan lokal, ya allah akhirnya selesai")
loaded_retriever = loaded_vectorstore.as_retriever()
print("Sistem RAG siap menggunakan database yang diload BOSSSSS!")
```

Untuk menguji apakah vektor database sudah berhasil disimpan

```python
kata_kunci = "negara mana saja yang termasuk blok poros atau axis?"
hasil_pencarian = vectorstore.similarity_search(kata_kunci, k=3)
print("---> HASIL PENCARIAN DI BUKU CATATAN <---")
for i, teks in enumerate(hasil_pencarian):
    print(f"\nCatatan ke-{i+1}:")
    print(teks.page_content)
```

### **Alasan Pemilihan Komponen Sistem RAG**

Dalam membangun sistem RAG ini, saya memilih beberapa komponen spesifik dengan pertimbangan sebagai berikut:

1. **Model Embedding (`paraphrase-multilingual-MiniLM-L12-v2` dari HuggingFace):** Saya memilih model ini karena mendukung pemrosesan banyak bahasa atau multilingual, termasuk bahasa Indonesia. Karena dokumen PDF *knowledge base* yang saya gunakan berbahasa Indonesia, model ini sangat ideal untuk menerjemahkan teks tersebut menjadi vektor yang akurat.

2. **Vector Database (Chroma DB):** Chroma dipilih karena merupakan *database* vektor *open-source* yang ringan, mudah diimplementasikan, dan berjalan sangat lancar di memori lokal seperti Google Colab untuk kebutuhan *mini-project*.

3. **Model Bahasa / LLM (`gemini-2.5-flash`):** Saya menggunakan Gemini 2.5 Flash karena model ini memiliki waktu respon yang sangat cepat dan gratis
digunakan melalui API. Selain itu, Gemini sangat pintar dalam memahami konteks bahasa Indonesia dan mampu menyusun kalimat jawaban yang luwes dan mudah dipahami oleh pengguna.
